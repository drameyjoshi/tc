# The goal
I want to calculate the electronic energies of Li atom using the Hartree-Fock (HF) method. HF computes the molecular orbitals - a molecule here means any arrangement of nuclei and electrons. Thus, in the context of HF theory, an atom is a molecule with one nucleus.

The model of Li atom that I learnt in elementary chemistry course has the two 1s electrons with the same energy. This is how the hydrogenic orbitals are. However, in multi-electron atoms, this is no longer true for the following reasons:

- In a true quantum picture of an atom, the terms 'inner' and 'outer' refer to energies, not locations.
- Electrons are indistinguishable. Therefore, one cannot label them as 'inner' and 'outer'. All one can say is that there is a certain occupancy in each orbital.
- We cannot apply the classical Gauss law and conclude that the 'outer' electrons do not affect the 'inner' electrons. Not because Gauss law is inapplicable at atomic scales but because there is no such thing as outer and inner electrons.

In the case of ground state of Li atom, the three electrons find themselves in the K and L shells. (Think in terms of energy, not orbit!) The $\alpha$ electron in the K shell will be exchange-correlated with the $\alpha$ electron of the L shell. Pauli exclusion principle prevents them from occupying the same space, which reduces their mutual Coulomb repulsion. However, there is no such exchange correlation between the $\beta$ electron in the K shell and the $\alpha$ electron in the L shell. As a result, the effective Coulomb repulsion experienced by the $K\beta$ electron is greater, which shows up as the $K\beta$ electron having a slightly greater (less negative) energy than the $K\alpha$ electron.

Strictly speaking, this conclusion should have been obvious with a qualitative reason and should have been only confirmed quantitatively. However, the simple model of atoms used to get started in chemistry are quite sticky.

# Gemini's explanation of exchange correlation

 Even though the $K\alpha$ ($1s$) and $L\alpha$ ($2s$) electrons are labeled with different orbital quantum numbers, the Pauli Exclusion Principle (PEP) operates on the entire wave function of indistinguishable fermions, not just on individual labels. It dictates that the total electronic wavefunction must be completely antisymmetric under the exchange of any two electrons.

Because they have the same spin (both are $\alpha$$\alpha$), this mathematical antisymmetry forces the spatial part of their wavefunction to vanish when their coordinates coincide ($x_1 = x_2$$x_1 = x_2$). This creates what is known as a Fermi hole (a region of depleted probability of finding the two electrons close to one another), which physically keeps them apart and reduces their mutual Coulomb repulsion.

Between a $K\beta$ and $L\alpha$ electron, the spins are different, so the total wavefunction's antisymmetry constraint does not restrict their spatial overlap in the same way (no Fermi hole is formed). This difference in spatial separation due to spin alignment is precisely what we call exchange correlation or exchange energy.

To see why this happens mathematically, let's look at the simplified wavefunctions for two electrons:

For $K\alpha$ and $L\alpha$ (same spin): Since both electrons have the same spin function $\alpha$$\alpha$, the spin part of the wavefunction is symmetric under exchange: $$\chi_{\text{spin}}(1,2) = \alpha(1)\alpha(2)$$$$\chi_{\text{spin}}(1,2) = \alpha(1)\alpha(2)$$

To satisfy the Pauli Exclusion Principle, the total wavefunction must be antisymmetric. Therefore, the spatial wavefunction must be antisymmetric: $$\Psi(1,2) = \frac{1}{\sqrt{2}} \big[ \psi_{1s}(1)\psi_{2s}(2) - \psi_{2s}(1)\psi_{1s}(2) \big] \alpha(1)\alpha(2)$$$$\Psi(1,2) = \frac{1}{\sqrt{2}} \big[ \psi_{1s}(1)\psi_{2s}(2) - \psi_{2s}(1)\psi_{1s}(2) \big] \alpha(1)\alpha(2)$$

If you evaluate this spatial part when the two electrons are at the exact same position in space ($r_1 = r_2 = r$$r_1 = r_2 = r$), you get: $$\Psi(r, r) \propto \psi_{1s}(r)\psi_{2s}(r) - \psi_{2s}(r)\psi_{1s}(r) = 0$$$$\Psi(r, r) \propto \psi_{1s}(r)\psi_{2s}(r) - \psi_{2s}(r)\psi_{1s}(r) = 0$$

The probability of finding them at the same spot is exactly zero! This spatial avoidance reduces their Coulomb repulsion.

For $K\beta$ and $L\alpha$ (opposite spin): Because their spins are different, we do not have a simple symmetric spin state. Instead, the Slater determinant naturally allows combinations where the spatial wavefunction does not have to vanish when $r_1 = r_2$$r_1 = r_2$ because the antisymmetry requirement is carried by the mixed spin state. Consequently, there is no spatial "Fermi hole" keeping them apart. They experience the full, unmitigated Coulomb repulsion.

Summary
The Pauli Exclusion Principle is not just a rule about "filling boxes" with different quantum numbers; it is a fundamental symmetry requirement on the multi-particle wavefunction. It is this requirement that actively shapes the spatial distribution of same-spin electrons (like $K\alpha$ and $L\alpha$), stabilizing them compared to opposite-spin pairs.

```python
import pyscf
from pyscf import gto, scf
```

The electronic configuration of Li is 1s$^2$2s. It has one unpaired electron. Therefore, its total spin $S = 1/2$ (in units of $\hbar$), making it a singlet state ($2S = 1$ ). It squared spin is $S^2 = S(S + 1) = 3/4$.

```python
atom = gto.M(
    atom="Li 0 0 0",
    basis="cc-pcvtz",
    charge=0,
    spin=1,
    verbose=3,  # Level of output (3 prints iteration progress)
)

print(f"Number of electrons: {atom.nelectron}")
print(f"Alpha electrons: {atom.nelec[0]}, Beta electrons: {atom.nelec[1]}")
print("-" * 50)
```

# Restricted Open-Shell HF

```python
rohf_res = scf.ROHF(atom)
rohf_energy = rohf_res.kernel()
print(f"Ground State Energy: {rohf_energy:.8f} Hartree")

rofh_ss, rohf_s = rohf_res.spin_square()
print(f"Ground State Spin Square: {rofh_ss:.4f}")
```

```python
# Print the molecular orbital (MO) energies
print("=== Molecular Orbital Energies (Hartrees) ===")
for i, energy in enumerate(rohf_res.mo_energy):
    # Determine occupation of the orbital (2 = fully occupied, 1 = half occupied, 0 = empty)
    occ = rohf_res.mo_occ[i]
    if occ > 0:
        print(f"Orbital {i+1:2d}: Energy = {energy:11.6f} Hartree | Occupation = {occ}")
    else:
        break
```

# Unrestricted HF

```python
uhf_res = scf.UHF(atom)
uhf_energy = uhf_res.kernel()
print(f"Ground State Energy: {uhf_energy:.8f} Hartree")

uhf_ss, uhf_s = uhf_res.spin_square()
print(f"Ground State Spin Square: {uhf_ss:.6f}")
print(f"Spin contamination: {uhf_ss - 0.75:.6f}")
```

```python
# Access alpha (spin-up) and beta (spin-down) orbital data
mo_energy_alpha, mo_energy_beta = uhf_res.mo_energy
mo_occ_alpha, mo_occ_beta = uhf_res.mo_occ

print("=== UHF Alpha (Spin-Up) Orbital Energies ===")
for i, energy in enumerate(mo_energy_alpha):
    occ = mo_occ_alpha[i]
    if occ > 0:
        print(f"Alpha Orbital {i+1:2d}: Energy = {energy:11.6f} Hartree | Occupation = {occ}")
    else:
        break

print("\n=== UHF Beta (Spin-Down) Orbital Energies ===")
for i, energy in enumerate(mo_energy_beta):
    occ = mo_occ_beta[i]
    if occ > 0:
        print(f"Beta Orbital {i+1:2d} : Energy = {energy:11.6f} Hartree | Occupation = {occ}")
    else:
        break

print("\n=== UHF loss of degeneracy in K shell ===")
print(f"Energy of alpha electron: {mo_energy_alpha[0]: 11.6f}")
print(f"Energy of beta electron:  {mo_energy_beta[0]: 11.6f}")
```

```python
# Access and print both occupied and virtual (unoccupied) orbitals around the HOMO-LUMO gap
mo_energy_alpha, mo_energy_beta = uhf_res.mo_energy
mo_occ_alpha, mo_occ_beta = uhf_res.mo_occ

print("=== UHF Alpha Orbitals (Occupied & lowest Virtuals) ===")
for i in range(5):
    occ = mo_occ_alpha[i]
    print(f"Alpha Orbital {i+1:2d}: Energy = {mo_energy_alpha[i]:11.6f} Hartree | Occupation = {occ}")

print("\n=== UHF Beta Orbitals (Occupied & lowest Virtuals) ===")
for i in range(5):
    occ = mo_occ_beta[i]
    print(f"Beta Orbital {i+1:2d} : Energy = {mo_energy_beta[i]:11.6f} Hartree | Occupation = {occ}")
```

# Comparison of RHF and UHF

```python
# Comparison to Hartree-Fock limit (~ -7.432727 Hartree)
hf_limit = -7.4327269
print("-" * 50)
print(f"Energy difference to numerical HF limit:")
print(f"   ROHF energy: {rohf_energy:11.6f}")
print(f"    UHF energy: {uhf_energy: 11.6f}")
print(f"      HF limit: {hf_limit: 11.6f}")
print(f"  ΔE (ROHF - Limit): {(rohf_energy - hf_limit) * 1000:.3f} mHartree")
print(f"  ΔE (UHF - Limit) : {(uhf_energy - hf_limit) * 1000:.3f} mHartree")
```

```python
from mendeleev import element

# Retrieve Lithium data
li = element('Li')

# Experimental Ionization Energies are typically reported in eV
# We convert eV to Hartree (1 Hartree ≈ 27.211386245988 eV)
ev_to_hartree = 1 / 27.211386245988

print("=== Experimental Ionization Energies for Lithium ===")
for degree, ie in li.ionenergies.items():
    print(f"IE {degree}+: {ie:8.4f} eV | {ie * ev_to_hartree:9.6f} Hartree")

# The sum of all ionization energies represents the total electronic binding energy
total_binding_energy_ev = sum(li.ionenergies.values())
total_binding_energy_hartree = total_binding_energy_ev * ev_to_hartree

print("\n=== Experimental Ground State Energy Approximation ===")
print(f"Total Binding Energy: {total_binding_energy_ev:.4f} eV")
print(f"Reconstructed Experimental Energy: -{total_binding_energy_hartree:.6f} Hartree")
```

#Comparison with experimental ionization energy

According to Koopmans' theorem, the energy to remove an electron from a specific orbital is approximately the negative of that orbital's energy.

```python
exp_ie1 = li.ionenergies[1] * ev_to_hartree  # Experimental 2s ionization

# ROHF energies
rohf_homo = rohf_res.mo_energy[1]

# UHF energies
uhf_homo_alpha = mo_energy_alpha[1]

print("=== Comparison for the Outer (2s) Electron Ionization ===")
print(f"Experimental 1st IE:                {exp_ie1:11.6f} Hartree")
print(f"ROHF HOMO Energy (-ε):              {-rohf_homo:11.6f} Hartree (Error: {abs(exp_ie1 - (-rohf_homo)):.5f})")
print(f"UHF Alpha HOMO Energy (-ε):         {-uhf_homo_alpha:11.6f} Hartree (Error: {abs(exp_ie1 - (-uhf_homo_alpha)):.5f})")

print("\n=== Summary ===")
print("Notice how the UHF Alpha HOMO energy (-0.196354 Hartree) is remarkably close")
print("to the experimental ionization energy (0.198142 Hartree). By allowing the alpha")
print("and beta spatial orbitals to polarize and separate (unrestricted), UHF captures")
print("the physical reality of electron interaction far better than ROHF, despite the ")
print("slight spin contamination.")
```

# Exact Numerical Solution via Full Configuration Interaction (FCI)
While we cannot solve the Schrödinger equation analytically, we can solve it numerically to the limit of our basis set using the Full Configuration Interaction (FCI) method, which exactly diagonalizes the Hamiltonian for the 3 electrons.

```python
from pyscf import fci

# Set up and run FCI using our existing UHF/ROHF reference
fci_solver = fci.FCI(atom, rohf_res.mo_coeff)
fci_energy, fci_coeffs = fci_solver.kernel()

print("=== Exact (FCI) vs. Mean-Field (HF) Energies ===")
print(f"ROHF Energy:       {rohf_energy:15.8f} Hartree")
print(f"UHF Energy:        {uhf_energy:15.8f} Hartree")
print(f"FCI Energy:        {fci_energy:15.8f} Hartree")
print(f"Experimental Est:  {-total_binding_energy_hartree:15.8f} Hartree")

# Correlation energy is the energy recovered beyond HF
corr_energy = fci_energy - rohf_energy
print(f"\nCorrelation Energy Recovered: {corr_energy * 1000:.3f} mHartree")
```

# Going Further: FCI with the larger `cc-pVQZ` Basis Set
To minimize the remaining basis set truncation error, we will rerun the ROHF and FCI calculations using the larger quadruple-zeta basis set `cc-pVQZ`.

```python
# Define the Lithium atom with the larger cc-pVQZ basis set
atom_vqz = gto.M(
    atom="Li 0 0 0",
    basis="cc-pVQZ",
    charge=0,
    spin=1,
    verbose=0
)

# Run the ROHF calculation first
rohf_vqz = scf.ROHF(atom_vqz)
rohf_vqz_energy = rohf_vqz.kernel()

# Run the FCI calculation using the ROHF orbitals
fci_vqz = fci.FCI(atom_vqz, rohf_vqz.mo_coeff)
fci_vqz_energy, _ = fci_vqz.kernel()

print("=== Comparison of Basis Sets: cc-pCVTZ vs cc-pVQZ ===")
print(f"cc-pCVTZ ROHF Energy: {rohf_energy:15.8f} Hartree")
print(f"cc-pVQZ  ROHF Energy: {rohf_vqz_energy:15.8f} Hartree")
print("-" * 50)
print(f"cc-pCVTZ FCI Energy:  {fci_energy:15.8f} Hartree")
print(f"cc-pVQZ  FCI Energy:  {fci_vqz_energy:15.8f} Hartree")
print(f"Experimental Est:     {-total_binding_energy_hartree:15.8f} Hartree")
print("-" * 50)
print(f"Remaining gap to experiment (cc-pCVTZ): {abs(fci_energy - (-total_binding_energy_hartree)) * 1000:.3f} mHartree")
print(f"Remaining gap to experiment (cc-pVQZ):  {abs(fci_vqz_energy - (-total_binding_energy_hartree)) * 1000:.3f} mHartree")
```

# The Correct Quadruple-Zeta Choice: `cc-pCVQZ`
To properly recover both core and valence correlation, we need the Core-Valence version of the quadruple-zeta basis set, `cc-pCVQZ`.

```python
# Define the Lithium atom with the correct cc-pCVQZ basis set
atom_cvqz = gto.M(
    atom="Li 0 0 0",
    basis="cc-pCVQZ",
    charge=0,
    spin=1,
    verbose=0
)

# Run the ROHF calculation first
rohf_cvqz = scf.ROHF(atom_cvqz)
rohf_cvqz_energy = rohf_cvqz.kernel()

# Run the FCI calculation using the ROHF orbitals
fci_cvqz = fci.FCI(atom_cvqz, rohf_cvqz.mo_coeff)
fci_cvqz_energy, _ = fci_cvqz.kernel()

print("=== Correct Comparison: cc-pCVTZ vs cc-pCVQZ ===")
print(f"cc-pCVTZ ROHF Energy: {rohf_energy:15.8f} Hartree")
print(f"cc-pCVQZ ROHF Energy: {rohf_cvqz_energy:15.8f} Hartree")
print("-" * 50)
print(f"cc-pCVTZ FCI Energy:  {fci_energy:15.8f} Hartree")
print(f"cc-pCVQZ FCI Energy:  {fci_cvqz_energy:15.8f} Hartree")
print(f"Experimental Est:     {-total_binding_energy_hartree:15.8f} Hartree")
print("-" * 50)
print(f"Remaining gap to experiment (cc-pCVTZ): {abs(fci_energy - (-total_binding_energy_hartree)) * 1000:.3f} mHartree")
print(f"Remaining gap to experiment (cc-pCVQZ): {abs(fci_cvqz_energy - (-total_binding_energy_hartree)) * 1000:.3f} mHartree")
```
